```python
import pandas as pd
import numpy as np

# 1. Simulación de carga de datos para Profiling & Diagnostic
print("=== 1. RAW DATA PROFILING & QUALITY DIAGNOSIS ===")

# Muestra representativa de dengue_abierto.csv
raw_data = {
    'FECHA_SIGN_SINTOMAS': ['2024-01-15', '2024-01-18', '2024-02-01', None, '2024-02-10'],
    'ENTIDAD_RES': [30, 30, 12, 12, 30],
    'MUNICIPIO_RES': [1, 193, 5, 5, 193],
    'ESTATUS_CASO': [2, 1, 2, 2, 2],  # 2: Confirmado, 1: Probable/No confirmado
    'EDAD': [25, 42, 15, 68, 31]
}

df_raw = pd.DataFrame(raw_data)

print("\n--- Raw Data Types & Null Counts ---")
print(df_raw.info())

print("\n--- Missing Value Summary ---")
print(df_raw.isnull().sum())

# 2. Aplicación de reglas de limpieza y estandarización (Data Cleaning)
print("\n=== 2. EXECUTING DATA CLEANING & STANDARDIZATION ===")
df_clean = df_raw.copy()

# Regla CLEAN-01: Filtrar solo casos confirmados (ESTATUS_CASO == 2)
df_clean = df_clean[df_clean['ESTATUS_CASO'] == 2]

# Regla CLEAN-02: Convertir fechas a Datetime
df_clean['FECHA_SIGN_SINTOMAS'] = pd.to_datetime(df_clean['FECHA_SIGN_SINTOMAS'])

# Regla CLEAN-03: Zero-padding para claves geográficas INEGI (5 dígitos)
df_clean['ENTIDAD_RES'] = df_clean['ENTIDAD_RES'].astype(str).str.zfill(2)
df_clean['MUNICIPIO_RES'] = df_clean['MUNICIPIO_RES'].astype(str).str.zfill(3)
df_clean['CVE_GEO'] = df_clean['ENTIDAD_RES'] + df_clean['MUNICIPIO_RES']

# Extraer semana y año epidemiológico
df_clean['EPI_YEAR'] = df_clean['FECHA_SIGN_SINTOMAS'].dt.isocalendar().year
df_clean['EPI_WEEK'] = df_clean['FECHA_SIGN_SINTOMAS'].dt.isocalendar().week

print("\n--- Cleaned & Processed Dataset Preview ---")
print(df_clean[['CVE_GEO', 'FECHA_SIGN_SINTOMAS', 'EPI_YEAR', 'EPI_WEEK', 'ESTATUS_CASO']])
```
